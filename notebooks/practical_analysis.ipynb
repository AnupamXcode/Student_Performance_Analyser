# Student Performance Statistical Analysis - Practical Notebook

This notebook demonstrates step-by-step statistical evaluation of student performance dataset covering:
1. **Practical 1**: Data Preprocessing & Descriptive Statistics
2. **Practical 2**: Probability Distributions (Uniform, Binomial, Poisson)
3. **Practical 3**: Sampling Techniques & Central Limit Theorem (CLT)
4. **Practical 4**: Non-Parametric Bootstrap Estimation

```python
import os
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

# Set seed for reproducibility
np.random.seed(42)
print("Libraries successfully imported!")
```

## 1. Data Cleaning & Descriptive Statistics

```python
# Load dataset
df = pd.read_csv('../data/student_performance.csv')
print(f"Raw dataset shape: {df.shape}")

# Clean duplicates and numerical columns
df_clean = df.drop_duplicates().reset_index(drop=True)
for col in ['Attendance', 'Study_Hours', 'Marks']:
    df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')
    df_clean.loc[df_clean[col] < 0, col] = np.nan
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())

print(f"Cleaned dataset shape: {df_clean.shape}")
df_clean.describe()
```

## 2. Probability Distributions Simulation

```python
# Binomial Distribution B(n=10, p=0.5)
n, p = 10, 0.5
binom_samples = np.random.binomial(n, p, size=1000)
prob_5_exact = stats.binom.pmf(5, n, p)
prob_5_observed = np.mean(binom_samples == 5)

print(f"Theoretical P(X=5): {prob_5_exact:.4f}")
print(f"Observed P(X=5): {prob_5_observed:.4f}")
```

## 3. Sampling Techniques & CLT Simulation

```python
# Central Limit Theorem Demo
pop_marks = df_clean['Marks'].values
pop_mean = np.mean(pop_marks)
pop_std = np.std(pop_marks, ddof=1)

sample_sizes = [10, 30, 50, 100]
plt.figure(figsize=(12, 8))

for i, n in enumerate(sample_sizes, 1):
    means = [np.mean(np.random.choice(pop_marks, size=n, replace=True)) for _ in range(1000)]
    plt.subplot(2, 2, i)
    sns.histplot(means, kde=True)
    plt.axvline(pop_mean, color='r', linestyle='--')
    plt.title(f"Sample Size N={n} | Observed SE: {np.std(means):.3f} vs Theoretical: {pop_std/np.sqrt(n):.3f}")

plt.tight_layout()
plt.show()
```

## 4. Non-Parametric Bootstrap Estimation

```python
# Bootstrap Resampling
initial_sample = np.random.choice(pop_marks, size=50, replace=False)
boot_means = [np.mean(np.random.choice(initial_sample, size=50, replace=True)) for _ in range(5000)]

boot_mean = np.mean(boot_means)
boot_se = np.std(boot_means, ddof=1)
ci_lower, ci_upper = np.percentile(boot_means, [2.5, 97.5])

print(f"True Population Mean: {pop_mean:.2f}")
print(f"Bootstrap Mean Estimate: {boot_mean:.2f}")
print(f"Bootstrap Standard Error: {boot_se:.4f}")
print(f"95% Confidence Interval: [{ci_lower:.2f}, {ci_upper:.2f}]")
```
